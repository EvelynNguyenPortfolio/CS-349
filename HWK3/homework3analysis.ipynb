# How does configuring SGDClassifier(loss='log_loss') differ from standard linear classification models, and what advantages do probability outputs and ROC-AUC metrics provide when evaluating customer churn?

SGDClassifier(loss='log_loss') differs from the standard linear classification model by updating model weights incrementally using single samples instead of evaluating the entire dataset all at once. Probability outputs enable companies to introduce granularity and churn risk distinctions or rankings among customers. They also allow companies to find the best middle ground where they can incentivize customers to prevent churning with the least amount of costs. ROC-AUC metrics measures the model's diagnostic quality across all operational thresholds simultaneously and provides meaningful model performance evaluation when dataset have class imbalance. Model performance can be misleading with class imbalance because a model can be highly accurate by just predicting the majority class and failing to detect the rare minority class.
