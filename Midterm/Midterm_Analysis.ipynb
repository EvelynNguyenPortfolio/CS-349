How did comparing StandardScaler versus MinMaxScaler impact your model training convergence, gradient descent optimization, and final evaluation metrics?

In terms of convergence, MinMaxScaler achieved faster training convergence (fewer iterations/epochs) than StandardScaler in both the Softmax Regression and the SGD Classifier. In the Softmax Pipeline,StandardScaler was 82 to MinMaxScaler was 74. In the SGDClassifier Pipeline, StandardScaler was 40 and MinMaxScaler 31. Reasons for this is because the MinMax shifts and rescales data so that all values fall within the bounded range of 0 to 1. Because the temporal feature extraction step already converted hours, months, and days into binary 0 and 1 columns, MinMaxScaler created numerical uniformity across the entire dataset, making it easier to optimize.

While the scaling method strongly impacted convergence speed, its impact on the final predictive performance was minimal. In the Softmax Performance, the StandardScaler had a better performance, achieving 79.81% accuracy compared to the 78.42% achieved by MinMaxScaler. Looking at the confusion matrices, StandardScaler was better at minimizing mistakes in the category (Class 1), correctly identifying 418 instances versus MinMaxScaler's 396. In the SGD Performance, the choice of scaler made virtually no difference for the stochastic engine. StandardScaler hit 75.53% accuracy, while MinMaxScaler hit 75.59%.
Across all 4 models, the precision and recall for Class 2 (Extreme Pollution) totaled 0.88 F1-score.
